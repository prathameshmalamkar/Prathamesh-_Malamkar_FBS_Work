```python
df=pd.read_csv("C:/Users/ASUS/Downloads/student_10.csv")
df
```

```python
df.info()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna()   #default axis=0 for row
```

```python
#numeric - median or mean
#categorical- mode
```

```python
med_per=df['Percentage'].median()
med_per
```

```python
df['Percentage']=df['Percentage'].fillna(med_per)
```

```python
mean_sh=df['Study hrs'].mean()
```

```python
df['Study hrs']=df['Study hrs'].fillna(mean_sh)
```

```python
df
```

```python
df['Student ID'].interpolate()
```

```python
df['Student ID'].bfill()
```

```python
df['Student ID'].ffill()
```
