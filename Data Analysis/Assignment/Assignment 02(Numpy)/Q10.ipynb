Q10. Create a 2x2 matrix A and a 2x2 matrix B. Perform matrix multiplication using np.dot() or the @ operator.


```python
import numpy as np
```

```python
arr1 = np.arange(4).reshape(2,2)
arr1
```

```python
arr2 = np.random.randint(1,10,(2,2))
arr2
```

```python
res =np.dot(arr1,arr2)
res
```

```python
res2 = arr1 @ arr2
res2
```
