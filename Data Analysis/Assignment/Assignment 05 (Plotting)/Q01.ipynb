### Refer to the tips.csv dataset. The dataset contains total 244 tips(data about customers served by a restaurant) across 7 different columns.



totbill - Total bill (cost of the meal), including tax, in US dollars

tip - Tip (gratuity) in US dollars

sex - Sex of person paying for the meal (0=male, 1=female)

smoker - Smoker in party? (0=No, 1=Yes)

day - Sun, Sat, Thu, Fri

time - Night and Day

size - Size of the party

1. Plot a day wise boxplot of totbill column.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv(r"C:/Users/ASUS/Downloads/tips.csv")
df
```

```python
sns.boxplot(x='day', y='total_bill', data=df)

plt.title('Day-wise Boxplot of Total Bill')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

2. Plot a pariplot by setting hue parameter t


```python
sns.pairplot(df, hue='sex')
plt.show()
```

3. Scatter plot of tip and totbill according to sex

```python
sns.scatterplot(
    x='total_bill',
    y='tip',
    hue='sex',
    data=df
)

plt.title('Tip vs Total Bill by Sex')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

4. Histogram of totbill for 50 bins


```python
sns.histplot(
    df['total_bill'],
    bins=50
)

plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

5. Heatmap for entire dataset's correlation


```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```
