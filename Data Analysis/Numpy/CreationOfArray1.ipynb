### Number of values inside reshape() = number of dimensions.
### Product of those values = total number of elements.

```python
arr3 =np.arange(1,43).reshape((6,7))
arr3
```

```python
arr3[[0,2,5]] # arr[[no.of each rows which you want]]
```

```python
arr3[:,[0,2,5]] # arr[[all rows : ,no of columns which you want]]
```

```python
arr3[:,-1] # arr[[:only last column]]   # all rows + ONLY last column
```

```python
arr3[1:4,1:4] #arr[[start rows:end row ,start column : end column]]
```

```python
arr3[:,:-1]  #not include last column  # all rows + all columns EXCEPT last
```
