## Bar Plot

1.Comparing categories

2.Showing frequencies and count

```python
products =["hot coffee","cold coffee","Tea","Salad"]
summer =[30,90,80,50]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(5,3))  # (width,height)
plt.bar(products,summer,color="purple",edgecolor="black",alpha=0.5,label="Summer")
plt.legend()
plt.title("sales of product for summer")
plt.xlabel("products")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.yticks(range(0, 101, 10))
# plt.grid(axis="y")
# plt.grid(axis="x")
plt.show()
```

```python
# Area chart
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.fill_between(x, y)

plt.show()
```

```python
fig, ax = plt.subplots(2, 2)
products=["laptop","mobile","phone","pc"]
sales=[40,20,50,80]
ax[0, 0].plot(x, y)
ax[0, 1].bar(products, sales)
ax[1, 0].scatter(x, y)
ax[1, 1].hist(y)
```
