### 1. Bar Chart:

a. Create a bar chart to compare the average price of cars by manufacturer.

a. Label the axes and add a suitable title.

b. Use different colors for different manufacturers.

```python
import matplotlib.pyplot as plt

avg_price = df.groupby('Manufacturer')['Price'].mean()

plt.bar(avg_price.index, avg_price.values,color=['red', 'blue', 'green', 'orange', 'purple'])

plt.xlabel('Manufacturer')
plt.ylabel('Average Price')
plt.title('Average Price of Cars by Manufacturer')
plt.xticks(rotation=45)

plt.show()
```

### 2. Line Chart:

a. Create a line chart showing the trend of horsepower for different types of cars.

b. Include proper legends, labels, and a title

```python
avg_hp = df.groupby('Type')['Horsepower'].mean()

plt.plot(avg_hp.index, avg_hp.values, marker='o', label='Horsepower')

plt.xlabel('Car Type')
plt.ylabel('Average Horsepower')
plt.title('Horsepower by Car Type')
plt.legend()

plt.show()
```

### 3. Pie Chart:

a. Create a pie chart showing the proportion of different car types in the dataset.

b. Include percentages in the chart.

```python
type_count = df['Type'].value_counts()

plt.pie(
    type_count.values,
    labels=type_count.index,
    autopct='%1.1f%%'
)

plt.title('Proportion of Different Car Types')

plt.show()
```

### 4. Scatter Plot:

a. Create a scatter plot to show the relationship between horsepower and MPG
(Miles Per Gallon).

b. Label the axes and add a title.

```python
plt.scatter(df['Horsepower'], df['MPG.city'])

plt.xlabel('Horsepower')
plt.ylabel('MPG')
plt.title('Horsepower vs MPG')

plt.show()
```
