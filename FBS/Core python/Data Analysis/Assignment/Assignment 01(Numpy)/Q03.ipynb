Q03. Perform the following operations on an array of mobile phones prices 6999, 7500,11999, 27899, 14999, 9999.


a. Create a 1d-array of mobile phones prices

```python
import numpy as np
```

```python
arr1 =np.array([6999,7500,11999,27899,14999,9999])
arr1
```

b. Convert this array to float type

```python
arr1 =arr1.astype(float)
arr1
```

c.Append a new mobile price of 13999

```python
arr1=np.append(arr1,13999)
arr1
```

d. Reverse the array

```python
arr1=arr1[: : -1]
arr1
```

e. Apply GST of 18%


```python
arr1 =arr1 +(arr1 * 18/100)
arr1
```

f. Sort array in descending order


```python
arr1 =np.sort(arr1)[::-1]
arr1
```

g. Calculate average price


```python
average = np.mean(arr1)
average
```

h. Difference between maximum and minimum price


```python
difference = np.max(arr1) - np.min(arr1)
difference
```
