```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].hist(df['Age'], bins=20)
ax[0].set_xlabel('Age')
ax[0].set_ylabel('Number of Passengers')
ax[0].set_title('Age Distribution')

ax[1].hist(df['Fare'], bins=20)
ax[1].set_xlabel('Fare')
ax[1].set_ylabel('Number of Passengers')
ax[1].set_title('Fare Distribution')

plt.tight_layout()
plt.show()
```

```python
sns.countplot(data=df, x='Survived', hue='Sex')

plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.title('Survival by Sex')
```

```python
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(15, 10))
axes = axes.flatten()

# Features to compare against 'Survived'
features = ['Age', 'Fare', 'Pclass', 'SibSp', 'Parch']

for i, col in enumerate(features):
    # Changed boxplot to barplot
    sns.barplot(data=df, x='Survived', y=col, ax=axes[i], palette='Set2', errorbar='sd') 
    # errorbar='sd' shows standard deviation instead of confidence interval (optional)
    
    axes[i].set_title(f'Average {col} vs. Survival', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Survived (0 = Died, 1 = Survived)')
    axes[i].set_ylabel(f'Average {col}')

# Hide the 6th empty subplot
axes[5].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
df['age_group'] = pd.cut(df['Age'], bins=[0, 12, 18, 40, 80], labels=['Child','Teen','Adult','Senior'])

sns.countplot(data=df, x='age_group', hue='Survived')
plt.title('Survival Count by Age Group')
plt.show()
```

```python
# Calculate survival percentage within each age group
age_survival = df.groupby('age_group')['Survived'].mean() * 100

# Plot as a bar chart
plt.figure(figsize=(6, 4))
age_survival.sort_values().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Survival Rate by Age Group (%)')
plt.ylabel('Survival Percentage')
plt.xlabel('Age Group')
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
