```python
s1 =pd.Series([10,20,30,40,'abc',30]) # hetrogenus , duplicate allowed
s1
```

```python
s1 = pd.Series([10,20,30,40],index=['a','b','c','d']) # customized index
s1
```

```python
s2 = pd.Series([30,40,50,60],index=['c','d','e','f'])
s2
```

```python
s1 + s2  # index should be matched
```
